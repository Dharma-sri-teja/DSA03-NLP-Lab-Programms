1. Tiny Spell Checker using Word Counts
Algorithm

```python
from collections import Counter
import re

text = """
python is easy to learn
python is a programming language
python is easy and powerful
programming is fun and easy
"""

words = re.findall(r'\w+', text.lower())
word_count = Counter(words)

def edits1(word):
    letters = 'abcdefghijklmnopqrstuvwxyz'

    splits = [(word[:i], word[i:]) for i in range(len(word) + 1)]

    deletes = [L + R[1:] for L, R in splits if R]

    replaces = [L + c + R[1:]
                for L, R in splits if R
                for c in letters]

    inserts = [L + c + R
               for L, R in splits
               for c in letters]

    return set(deletes + replaces + inserts)

word = input("Enter a word: ").lower().strip()

if word in word_count:
    print("Spelling is correct")
else:
    suggestions = edits1(word)
    possible = [w for w in suggestions if w in word_count]

    if possible:
        correct = max(possible, key=lambda w: word_count[w])
        print("Spelling is incorrect")
        print("Did you mean:", correct)
        print("Probability/Frequency:", word_count[correct])
    else:
        print("Spelling is incorrect")
        print("No suggestion found")
```

2. Paragraph Analysis

```python
paragraph = input("Enter a paragraph: ")

sentences = paragraph.split(".")

print("\nSentences:")
for sentence in sentences:
    if sentence.strip():
        print(sentence.strip())

words = paragraph.split()

print("\nWords:")
print(words)

print("\nTotal number of words:", len(words))

count = {}

for word in words:
    word = word.lower().strip(".,!?")

    if word in count:
        count[word] += 1
    else:
        count[word] = 1

print("\nWord Frequency:")
for word in count:
    print(word, ":", count[word])

highest = max(count.values())

print("\nMost frequent words:")
for word in count:
    if count[word] == highest:
        print(word, ":", count[word])
```

3. Regular Expression Program

```python
import re

text = """Call me on 9876543210 or 9123456780.
My email is student@gmail.com and teacher@yahoo.com.
The meeting is on 25-12-2026.
Python and NLP are useful Technologies."""

date = re.search(r'\d{2}-\d{2}-\d{4}', text)

if date:
    print("First date:", date.group())

mobiles = re.findall(r'\b[6-9]\d{9}\b', text)

print("\nMobile numbers:")
print(mobiles)

emails = re.findall(r'[\w.-]+@[\w.-]+\.\w+', text)

print("\nEmail IDs:")
print(emails)

capitalized = re.findall(r'\b[A-Z][a-zA-Z]*\b', text)

print("\nCapitalized words:")
print(capitalized)
if re.match(r'call', text, re.IGNORECASE):
    print("\nText starts with 'call'")
else:
    print("\nText does not start with 'call'")
def change_date(match):
    day, month, year = match.group().split("-")
    return year + "/" + month + "/" + day

new_text = re.sub(r'\d{2}-\d{2}-\d{4}', change_date, text)

print("\nText after changing date:")
print(new_text)
hidden_text = re.sub(r'\b[6-9]\d{9}\b', 'XXXXXXXXXX', new_text)

print("\nText after hiding phone numbers:")
print(hidden_text)
```

4. Regular Expression Operators

```python
import re

patterns = {
    "?  Zero or one": r"ab?c",
    "*  Zero or more": r"ab*c",
    "+  One or more": r"ab+c",
    "[] Character set": r"[abc]+",
    "[^] Not in set": r"[^abc]+",
    "^  Starts with": r"^Hello",
    "$  Ends with": r"world$"
}

samples = {
    "?  Zero or one": ["ac", "abc", "abbc"],
    "*  Zero or more": ["ac", "abc", "abbbc"],
    "+  One or more": ["ac", "abc", "abbbc"],
    "[] Character set": ["abc", "cab", "xyz"],
    "[^] Not in set": ["xyz", "hello", "abc"],
    "^  Starts with": ["Hello world", "hello world", "Say Hello"],
    "$  Ends with": ["hello world", "world", "world hello"]
}

for name in patterns:
    print("\n", name)
    print("Pattern:", patterns[name])

    for text in samples[name]:
        if re.search(patterns[name], text):
            print(text, "-> YES")
        else:
            print(text, "-> NO")
```

5. Finite State Automaton — Ends with "ab"

```python
def check_string(string):
    state = "q0"

    print("\nInput:", string)
    print("Path:", state, end=" ")

    for ch in string:

        if state == "q0":
            if ch == "a":
                state = "q1"
            else:
                state = "q0"

        elif state == "q1":
            if ch == "a":
                state = "q1"
            else:
                state = "q2"

        elif state == "q2":
            if ch == "a":
                state = "q1"
            else:
                state = "q0"

        print("->", state, end=" ")

    if state == "q2":
        print("\nACCEPTED")
    else:
        print("\nREJECTED")


strings = ["ab", "aab", "abab", "baab", "a", "aba", "abb"]

for string in strings:
    check_string(string)
```
