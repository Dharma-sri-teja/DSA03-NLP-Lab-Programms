1. Regular Expressions – Match and Search Patterns

```python
import re

text = "My email is student@gmail.com and my phone is 9876543210."
email = re.search(r'\w+@\w+\.\w+', text)
phone = re.search(r'\d{10}', text)

if email:
    print("Email found:", email.group())
else:
    print("Email not found")

if phone:
    print("Phone number found:", phone.group())
else:
    print("Phone number not found")
```

2. Finite State Automaton – Strings Ending with ab

```python
def fsa(string):
    state = 0
    for ch in string:
        if state == 0:
            if ch == 'a':
                state = 1
            else:
                state = 0
        elif state == 1:
            if ch == 'b':
                state = 2
            elif ch == 'a':
                state = 1
            else:
                state = 0
        elif state == 2:
            if ch == 'a':
                state = 1
            else:
                state = 0
    return state == 2
s = input("Enter a string: ")
if fsa(s):
    print("ACCEPTED")
else:
    print("REJECTED")
```

3. Morphological Analysis using NLTK

```python
import nltk

from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

words = ["playing", "played", "studies", "cars"]

for word in words:
    print("Word:", word)
    print("Stem:", stemmer.stem(word))
    print("Lemma:", lemmatizer.lemmatize(word))
    print()
```

4. Finite-State Machine for Morphological Parsing – Plural Nouns

```python
def plural(noun):
    if noun.endswith("s"):
        return noun + "es"
    elif noun.endswith("x") or noun.endswith("ch") or noun.endswith("sh"):
        return noun + "es"
    else:
        return noun + "s"


words = ["cat", "dog", "box", "bus", "brush"]

for word in words:
    print(word, "->", plural(word))
```
